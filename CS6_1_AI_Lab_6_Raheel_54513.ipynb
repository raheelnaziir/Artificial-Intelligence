```python
import numpy as np

def evaluate_misplaced(puzzle, goal):
    count = 0
    for i in range(9):
        if puzzle[i] != goal[i] and puzzle[i] != 0:
            count += 1
    return count


def print_grid(puzzle):
    for i in range(0, 9, 3):
        print(puzzle[i], puzzle[i+1], puzzle[i+2])
    print()


def solve(puzzle, goal):

    steps = [
        ('up', -3),
        ('down', 3),
        ('left', -1),
        ('right', 1)
    ]

    visited = []
    priority = []

    h = evaluate_misplaced(puzzle, goal)
    priority.append((puzzle,0,h))

    visited.append(puzzle)

    while priority:

        priority = sorted(priority, key=lambda x: x[1]+x[2])
        puzzle, g, h = priority.pop(0)

        print("Current State:")
        print_grid(puzzle)

        if puzzle == goal:
            print("Goal Reached!")
            return

        blank = puzzle.index(0)

        for step in steps:

            move, pos = step
            new_blank = blank + pos

            if new_blank < 0 or new_blank >= 9:
                continue

            new_puzzle = puzzle.copy()
            new_puzzle[blank], new_puzzle[new_blank] = new_puzzle[new_blank], new_puzzle[blank]

            if new_puzzle not in visited:

                visited.append(new_puzzle)

                g_new = g + 1
                h_new = evaluate_misplaced(new_puzzle, goal)

                priority.append((new_puzzle,g_new,h_new))


# Initial State
puzzle = [2,8,3,
          1,6,4,
          7,0,5]

# Goal State
goal = [1,2,3,
        8,0,4,
        7,6,5]

solve(puzzle,goal)
```

#Implement A* example other than above example. i.e shortest path of graphs, maps, games.

```python
import heapq

# Graph representation
graph = {
    'A': [('B',1), ('C',4)],
    'B': [('D',3), ('E',2)],
    'C': [('F',5)],
    'D': [],
    'E': [('F',1)],
    'F': []
}

# Heuristic values (estimated distance to goal)
heuristic = {
    'A': 5,
    'B': 3,
    'C': 4,
    'D': 2,
    'E': 1,
    'F': 0
}

def astar(start, goal):

    open_list = []
    heapq.heappush(open_list, (0, start))

    g_cost = {node: float('inf') for node in graph}
    g_cost[start] = 0

    parent = {start: None}

    while open_list:

        current = heapq.heappop(open_list)[1]

        if current == goal:
            break

        for neighbor, cost in graph[current]:

            new_cost = g_cost[current] + cost

            if new_cost < g_cost[neighbor]:

                g_cost[neighbor] = new_cost
                f = new_cost + heuristic[neighbor]

                heapq.heappush(open_list, (f, neighbor))
                parent[neighbor] = current

    # Reconstruct path
    path = []
    node = goal

    while node is not None:
        path.append(node)
        node = parent[node]

    path.reverse()

    print("Shortest Path:", path)
    print("Total Cost:", g_cost[goal])


# Run A*
astar('A','F')
```

## We are implementing List in Python


```python
nameList = ["Raheel", "Nouman"]
print(type(nameList))
print(nameList)
print(nameList[0])
```

```python
nameList[1] #Use index to access the Value in List
```

```python
length = len(nameList)
for i in range(length):
  print(nameList[i])
```

```python
list = [10, 20 ,30, 40, 50]
print("#append()") #append
list.append(60)
print(list)

print("#clear") #clear
list.clear()
print(list)

print("#copy") #copy
list.copy()
print(list)

print("#count") #count
print(list.count(10))

print("#extend") #extend
fruit = ["Apple", "Banana", "Grapes"]
list.extend(fruit)
print(list)

print("#index") #index
print(list.index("Apple"))

print("#insert") #insert
list.insert(3, "Pomo")
print(list)

print("#pop") #pop
list.pop(3)
print(list)

print("#remove") #remove
list.remove("Apple")
print(list)

print("#reverse") #reverse
list.reverse()
print(list)

print("#sort") #sort
even = [10,8,6,4,2,0]
even.sort()
print(even)

#looping through an array

for i in even:
  print(i)
```
